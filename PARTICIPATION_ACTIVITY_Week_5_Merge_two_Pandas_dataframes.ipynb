```python
dfs_with_ticker=[]
for file in CSV_FILES:
    df=pd.read_csv(file)
    ticker_symbol=file.replace('.csv','').upper()
    df['ticker']=ticker_symbol
    dfs_with_ticker.append(df) #add column ticker
```

```python
all_stocks=pd.concat(dfs_with_ticker,ignore_index=True)
all_stocks.to_csv('faanf.csv',index=False)
df_fannf=pd.read_csv('faanf.csv')
df_fannf['date']=pd.to_datetime(df_fannf['date'])
df_fannf['volume']=df_fannf['volume'].astype(int)
df_fannf_sorted=df_fannf.sort_values(by=['date','ticker'])
```

```python
top7volume=df_fannf_sorted.nlargest(7,'volume')
print("Highest 7 volume：")
display(top7volume)
```

```python
melted_df=pd.melt(df_fannf,
    id_vars=['date','ticker'],
    var_name='metric',
    value_name='value')
print("melted data：")
display(melted_df.head())
```
