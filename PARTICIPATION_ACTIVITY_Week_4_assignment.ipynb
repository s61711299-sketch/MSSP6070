a. Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'

```python
import pandas as pd
```

```python
df=pd.read_csv('/content/Working with Pandas DataFrame/parsed.csv')
```

```python
japan_mb=df[df['place'].str.contains('Japan')&(df['magType']=='mb')]
JPpercentile_95=japan_mb['mag'].quantile(0.95)
print(JPpercentile_95)
```
