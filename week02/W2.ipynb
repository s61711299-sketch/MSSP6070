```python
letter = ['a','b','c','d']
print(int(int('3'*2)//11))
# 3 as integer, so '3'*2 ->33,33/11=3
print(letter[-1])
# [-1] means on the right side, so d
print(letter[:2])
# [:2] means the first two on the left side, so ['a','b']
```

```python
spam = ['apples','bananas','tofu','cats']
def results(spam):
  if len(spam) == 0:
    return ''
  elif len(spam) == 1:
    return str(spam[0])
  return ', '.join(spam[:-1])+f', and {spam[-1]}'

spam = ['apples','bananas','tofu','cats']
print(results(spam))
```
